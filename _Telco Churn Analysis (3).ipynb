```python
# replacing blanks with 0 as tenure is 0 and no total charges are recorded.
df['TotalCharges'] = df['TotalCharges'].replace(' ',0)
# converting the datatype 
df['TotalCharges'] = df['TotalCharges'].astype('float') 
```

```python
# converted values of SeniorCitizen column as 0 and 1 to yes/no to make it easier to understand.

def conv(value):
    if value == 1:
        return 'yes'
    else:
        return 'no'

df['SeniorCitizen'] = df['SeniorCitizen'].apply(conv)
```

```python
ax = sns.countplot(x = df['Churn'], data = df, edgecolor = 'k')
ax.bar_label(ax.containers[0])
plt.title('Customer count by churn' , size = 20, weight = 'bold', color = 'green', style = 'oblique')
plt.xlabel('Churn', weight = 'bold')
plt.ylabel('Count', weight = 'bold')
plt.show()
```

```python
gb = df.groupby('Churn').agg({'Churn' : 'count'})
plt.figure(figsize=(8,5))
gb['Churn'].plot.pie(labels = gb.index, autopct = '%1.1f%%' , legend = 'Churn')
plt.title('Percentage of Churned Customers', size = 20, weight = 'bold', color = 'green', style = 'oblique')
plt.xlabel('Churn', weight = 'bold')
plt.ylabel('Count', weight = 'bold')
plt.show()
```

### The pie chart indicates that 26.5% of our customers have churned. 


### Let’s analyze the underlying reasons for this churn to gain better insights.








```python
plt.figure(figsize=(4,4))
sns.countplot(x = 'gender',hue = 'Churn',data = df, edgecolor = 'k')
plt.title('Churn By Gender', size = 22, weight = 'bold', color = 'green', pad = 15, style ='oblique')
plt.xlabel('Gender', weight = 'bold')
plt.ylabel('Count', weight = 'bold')
plt.show()
```

```python
plt.figure(figsize=(4,5))
ax = sns.countplot(x = 'SeniorCitizen',data = df, edgecolor = 'k')
ax.bar_label(ax.containers[0])
plt.title('Count of Customers By Senior Citizen', size = 18, weight = 'bold', color = 'green', pad = 15, style ='oblique')
plt.xlabel('Senior Citizen', weight = 'bold')
plt.ylabel('Count', weight = 'bold')
plt.show()
```

```python
# Creating the stacked bar chart with percentages
stacked_data = df.groupby(['SeniorCitizen', 'Churn']).size().unstack(fill_value=0)
stacked_data_percentage = stacked_data.div(stacked_data.sum(axis=1), axis=0) * 100

stacked_data_percentage.plot(kind='bar', stacked=True, figsize=(6, 6), color=sns.color_palette('pastel'))

# Adding percentage labels on each bar
for i, row in stacked_data_percentage.iterrows():
    bottom = 0
    for j, value in enumerate(row):
        plt.text(i, bottom + value / 2, f'{value:.1f}%', ha='center', va='center', fontsize=9, color='black')
        bottom += value

plt.title('Churn By Senior Citizen (Percentage)', size=15, weight='bold', color='green', style ='oblique')
plt.xlabel('Senior Citizen', weight='bold')
plt.ylabel('Percentage', weight='bold')
plt.tight_layout()
plt.show()
```

### Comparitively, a greater percentage of people in the senior citizen category have churned.








```python
plt.figure(figsize=(11,5))
sns.histplot( x = 'tenure', data = df, bins = 72, hue = 'Churn')
plt.xlabel('Tenure', weight = 'bold')
plt.ylabel('Count', weight = 'bold')
plt.show()
```

### Long-term customers are more likely to stay, while those with a tenure of just one or two months are more likely to churn

```python
plt.figure(figsize=(7,5))
ax = sns.countplot(x = 'Contract',data = df, edgecolor = 'k', hue ='Churn')
ax.bar_label(ax.containers[0])
plt.title('Count of Customers By Contract', size = 20, weight = 'bold', color = 'green', pad = 15, style ='oblique')
plt.xlabel('Senior Citizen', weight = 'bold')
plt.ylabel('Count', weight = 'bold')
plt.show()
```

### People who have month to month contract are likely to churn than from those who have 1 or 2 years of contract

```python
# Define the columns to plot
columns = ['PhoneService', 'MultipleLines', 'InternetService', 
           'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
           'TechSupport', 'StreamingTV', 'StreamingMovies']

# Set the number of rows and columns for subplots
n_cols = 3
n_rows = -(-len(columns) // n_cols)  # Ceiling division to determine rows

# Create the subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15,10))
axes = axes.flatten()  # Flatten the 2D array of axes into 1D for easy iteration

# Generate countplots for each column
for i, col in enumerate(columns):
    sns.countplot(x = col, data = df, hue = df['Churn'], palette = 'pastel', ax = axes[i])
    axes[i].set_title(f'{col}', fontsize=14, color = 'green', style ='oblique')
    axes[i].set_xlabel('', weight = 'bold')
    axes[i].set_ylabel('Count', fontsize=10, weight = 'bold')
plt.tight_layout()
plt.show()
```

### The majority of customers who do not churn tend to have services like PhoneService, InternetService (particularly DSL), and OnlineSecurity enabled. For services like OnlineBackup, TechSupport, and StreamingTV, churn rates are noticeably higher when these services are not used or are unavailable.
